## Higher Education Trends Analysis

This project explore patterns of international student mobility from 2019 to 2023.Using visualizations the project highlights the most popular migration routes, preferred study disciplines, and trends in student migration over time.This shows where students go to study, which courses and countries are popular, and how they do after graduation. It helps universities and policymakers plan and support international students better.

```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns 
```

```python
plt.figure(figsize=(10,6))
top_origins = df["origin_country"].value_counts().head(10)
sns.barplot(x=top_origins.index,y=top_origins.values)
plt.title("Top 10 Origin Countries")
plt.xlabel("Country")
plt.ylabel("Number of Student")
plt.show()
```

```python
plt.figure(figsize=(10,6))
top_destinations = df["destination_country"].value_counts().head(10)
sns.barplot(x=top_destinations.index, y=top_destinations.values)
plt.title("Top 10 Destination Countries")
plt.xlabel("Country")
plt.ylabel("Number of Students")
plt.show()
```

```python
field=df["field_of_study"].value_counts()
plt.pie(field,labels=field.index, autopct="%1.1f%%")
plt.title("Distribution of field of Study")
plt.show()
```

```python
top_routes = df.groupby(["origin_country", "destination_country"]) \
               .size() \
               .sort_values(ascending=False) \
               .head(5)

top_routes.plot(kind="bar")
plt.title("Top 5 Student Migration Routes")
plt.xlabel("Origin → Destination")
plt.ylabel("Number of Students")
plt.xticks(rotation=45)
plt.show()
```

```python
yearly = df.groupby('year_of_enrollment')['student_id'].count().reset_index()
plt.figure(figsize=(6,4))
sns.lineplot(data=yearly, x='year_of_enrollment', y='student_id')
plt.title("Annual Trend in Student Migration (2019–2023)")
plt.xlabel("Year")
plt.ylabel("Total Students")
plt.show()
```

```python
counts = df['scholarship_received'].value_counts()
# Explode the first slice
explode = [0.1, 0]  
plt.figure(figsize=(8,6))
plt.pie(counts,labels=counts.index,autopct='%1.1f%%',explode=explode,startangle=90)
plt.title("Scholarship Status Distribution")
plt.axis('equal')
plt.show()
```

```python
plt.figure(figsize=(14,7))
sns.boxplot(data=df, x='destination_country', y='starting_salary_usd')
plt.xticks(rotation=45)
plt.title("Salary Distribution by Destination Country")
plt.show()
```

```python
numeric_cols = [
    'year_of_enrollment',
    'graduation_year',
    'starting_salary_usd',
    'gpa_or_score',
    'test_score',
]
df_numeric = df[numeric_cols]
corr = df_numeric.corr()
plt.figure(figsize=(10, 8))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Heatmap of Student Dataset')
plt.show()
```
